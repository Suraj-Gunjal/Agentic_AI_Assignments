# Assignment 4 — Finite State Machine Chatbot

**Objective:** Construct a finite state machine (FSM) chatbot with **Greeting, Help, and Goodbye** states.

The chatbot changes its state depending on what the user types. This is a simple example of how an agent can use a fixed set of states and rules to respond.

### How this chatbot works

The chatbot has three main states:

- **GREETING** → starts the conversation and welcomes the user.
- **HELP** → provides information when the user asks for help.
- **GOODBYE** → ends the conversation.

The important idea is that the **current state + user input determines the next state**.

````python
# Define the states used by our chatbot
GREETING = "greeting"
HELP = "help"
GOODBYE = "goodbye"
````

### State transition rules

We keep the rules simple so they are easy to understand:

- Greeting words such as `hello` or `hi` move the chatbot to the **Greeting** state.
- `help` moves it to the **Help** state.
- `bye`, `goodbye`, or `exit` move it to the **Goodbye** state.
- Anything else keeps the chatbot in its current state and gives a default response.

````python
def get_next_state(user_input, current_state):
    """Decide the next state based on the user's message."""

    text = user_input.lower().strip()

    if text in ["bye", "goodbye", "exit", "quit"]:
        return GOODBYE

    if text in ["help", "help me", "i need help"]:
        return HELP

    if text in ["hello", "hi", "hey", "good morning", "good evening"]:
        return GREETING

    # If there is no matching rule, stay in the same state
    return current_state
````

### Responses for each state

Each state has its own response. This keeps the FSM structure clear instead of putting all responses inside one large function.

````python
def get_response(state):
    if state == GREETING:
        return "Hello! Nice to meet you. How can I help you?"

    elif state == HELP:
        return "Sure! I can respond to greetings, provide help, or end the conversation."

    elif state == GOODBYE:
        return "Goodbye! Have a great day."

    return "I am not sure how to respond to that. Try typing 'help'."
````

### Run the chatbot

The conversation continues until the chatbot enters the **GOODBYE** state.

Try different inputs such as `hello`, `help`, `bye`, or something unrelated to see how the state changes.

````python
def run_chatbot():
    current_state = GREETING

    print("Chatbot: Hello! Type 'help' if you need assistance.")

    while current_state != GOODBYE:
        user_input = input("You: ")

        # Find the next state using the user's input
        current_state = get_next_state(user_input, current_state)

        print("Chatbot:", get_response(current_state))

    print("\nConversation ended.")


run_chatbot()
````

## Example conversation

A sample run can look like this:

The chatbot does not generate completely new behavior. It follows predefined transitions between states, which is the main idea of a **finite state machine**.

## Quick explanation for the lecture

**What is an FSM?**  
A finite state machine is a model where a system can be in one state at a time and changes state when a particular input or event occurs.

**Why is this chatbot an FSM?**  
Because it has a fixed number of states and predefined rules that decide which state comes next.

**What are the inputs?**  
The user's messages, such as `hello`, `help`, and `bye`.

**What are the outputs?**  
The chatbot's response for the current state.

**What ends the chatbot?**  
When the state becomes `GOODBYE`.

### Important point

This is intentionally a **simple rule-based FSM**, not a large language model. Its purpose is to demonstrate the concept of states and transitions clearly.
